```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/network_incidents.csv")

df.head()
```

```python
df.info()
```

```python
df["Incident_Type"].value_counts()
```

```python
df["Incident_Type"].value_counts().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Network Incident Frequency")

plt.xlabel("Incident Type")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
df["Severity"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6, 6)
)

plt.title("Severity Distribution")

plt.ylabel("")

plt.show()
```
